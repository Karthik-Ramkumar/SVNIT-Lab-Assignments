### 6. Define a roux array as a 1-D array such that, when it's reversed, it represents the sequence of square numbers 1,4,9,16,... with 0s Interwoven between them.

````python
import numpy as np

def create_roux_array(n):
    # Generate enough square numbers
    num_squares = (n + 1) // 2
    squares = [i**2 for i in range(1, num_squares + 1)]
    
    # Interweave with 0s
    seq = []
    for sq in squares:
        seq.append(sq)
        seq.append(0)
    
    # Trim to length n
    seq = seq[:n]
    
    # Reverse to get roux array
    roux = seq[::-1]
    return np.array(roux)

print("# roux array of length 5")
print(create_roux_array(5))

print("# roux array of length 8")
print(create_roux_array(8))
````

### 10. Write a program to count the occurrence of a specified item in a given NumPy array. And write a NumPy program to remove nan values from a given array.

````python
import numpy as np

# Part 1: count occurrence
arr_count = np.array([10, 20, 20, 20, 20, 0, 20, 30, 30, 30, 0, 0, 20, 20, 0])
print("Original array:")
print(arr_count)
unique, counts = np.unique(arr_count, return_counts=True)
print("Occurrences:", counts[np.argsort(unique)])

# Part 2: remove nan
arr_nan = np.array([200., 300., np.nan, np.nan, np.nan, 700.])
print("Original array:")
print(arr_nan)
print("After removing nan values:", arr_nan[~np.isnan(arr_nan)])
````

### 16. Find all the peaks in a 1D numpy array a. Peaks are points surrounded by smaller values on both sides.

````python
import numpy as np

a = np.array([1, 3, 7, 1, 2, 6, 0, 1])
peaks = np.where((a[1:-1] > a[:-2]) & (a[1:-1] > a[2:]))[0] + 1
print(peaks)
````
