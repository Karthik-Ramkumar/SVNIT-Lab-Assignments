### Methodology Overview:
In real-world data science applications (e.g., IoT sensor telemetry, financial trading streams, network traffic logs), data arrives continuously at high velocities. Storing or processing every incoming data point can be computationally prohibitive.
- **Sampling** reduces the volume of the data stream by picking a representative subset without evaluating the content of individual elements.
- **Filtering** inspects each incoming element against a predefined condition and retains only data points of interest, discarding the rest.


### 1. Import Required Libraries
Before processing the stream, we import the necessary Python libraries:
- `pandas`: For creating structured tables and analyzing retention metrics.
- `numpy`: For numerical computations and statistical operations.
- `random`: Python's built-in module for simulating the stream and performing random sampling.


```python
# Import required libraries
import pandas as pd
import numpy as np
import random

# Set random seed for reproducibility in laboratory evaluations
random.seed(42)
np.random.seed(42)

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)
print("Libraries imported successfully.")
```

### 2. Generate Simulated Data Stream
A data stream is an incoming sequence of items. Here, we simulate a stream of 100 random numerical values between 1 and 100 using Python's `random.randint()`. The values are stored in a standard Python list.


```python
# Generate a stream of 100 random numerical values between 1 and 100
stream_data = [random.randint(1, 100) for _ in range(100)]

# Display the original data stream
print("Original Data Stream (100 elements):")
print(stream_data)

# Display the total number of elements
print("\nTotal number of elements in the stream:", len(stream_data))
```

### 3. Apply Random Sampling
**Sampling** selects a subset of elements from the data stream. It is used to reduce the data volume for downstream analysis without inspecting or biasing towards specific values.

Here, we randomly select 20 values from the original stream using `random.sample()`.


```python
# Randomly select 20 values from the original stream using random.sample()
sample_size = 20
sampled_stream = random.sample(stream_data, sample_size)

# Display the sampled values
print(f"Sampled Stream ({sample_size} randomly selected values):")
print(sampled_stream)

# Display the sample size
print("\nSample size:", len(sampled_stream))
```

### 4. Apply Filtering
**Filtering** evaluates a logical condition (predicate) on each data element in the stream. Only elements that satisfy the condition are retained.

Here, we filter the original data stream to retain only values greater than 50 using Python list comprehension.


```python
# Filter the original data stream to retain only values greater than 50
filtered_stream = [x for x in stream_data if x > 50]

# Display the filtered values
print("Filtered Stream (Values > 50):")
print(filtered_stream)

# Display the number of filtered elements
print("\nNumber of filtered elements:", len(filtered_stream))
```

### 5. Analyze the Results
We now compare:
- Original stream size
- Sampled stream size
- Filtered stream size
- Percentage of original data retained after sampling
- Percentage of original data retained after filtering

We organize these results in a structured Pandas DataFrame.


```python
# Calculate stream sizes
original_size = len(stream_data)
sampled_size = len(sampled_stream)
filtered_size = len(filtered_stream)

# Calculate retention percentages
sampling_retention_pct = (sampled_size / original_size) * 100
filtering_retention_pct = (filtered_size / original_size) * 100

# Create comparison DataFrame using Pandas
comparison_df = pd.DataFrame({
    "Stream Operation": [
        "Original Stream",
        "Random Sampling (n = 20)",
        "Conditional Filtering (Value > 50)"
    ],
    "Element Count": [
        original_size,
        sampled_size,
        filtered_size
    ],
    "Retention Rate (%)": [
        "100.00%",
        f"{sampling_retention_pct:.2f}%",
        f"{filtering_retention_pct:.2f}%"
    ],
    "Selection Basis": [
        "Complete stream",
        "Random selection (independent of value)",
        "Condition check (value > 50)"
    ]
})

print("=== Stream Processing Comparison Summary ===")
comparison_df
```

### 6. Sampling vs. Filtering: Comparison

Both sampling and filtering reduce the volume of data flowing through a stream processing pipeline, but they do so based on fundamentally different mechanisms:

#### What Sampling Does:
- **Random / Probabilistic Selection**: It selects an arbitrary subset of items without evaluating the contents or values of individual data points.
- **Volume Reduction**: Its primary purpose is to cut down bandwidth, storage, and CPU consumption while maintaining an unbiased statistical representation of the data.
- **Predictable Output Size**: The size of the output is directly controlled by the user (e.g., exactly 20 items or 20% of the stream).

#### What Filtering Does:
- **Predicate-Based Selection**: It inspects each arriving item and tests it against a specific condition (e.g., `value > 50` or `status == "ERROR"`).
- **Relevance Extraction**: Its primary purpose is to isolate critical items, alerts, or anomalies from uninteresting background data.
- **Dynamic Output Size**: The number of retained items is not predetermined; it depends entirely on how many elements satisfy the condition.

#### Comparison Table:
| Feature / Aspect | Stream Sampling | Stream Filtering |
| :--- | :--- | :--- |
| **Primary Goal** | Reduce overall volume while preserving statistical properties | Extract relevant, actionable, or critical items |
| **Selection Basis** | Random / uniform probability (content-agnostic) | Logical condition / threshold (content-aware) |
| **Output Size** | Fixed and predetermined | Variable (depends on stream contents) |
| **Preservation of Critical Events**| Might miss rare anomalies due to random omission | Guarantees all matching events are captured |
| **Typical Use Cases** | Load shedding, telemetry approximation, trend monitoring | Fraud detection, alarm triggers, error log extraction |


### 7. Conclusion

In this experiment, we implemented stream data sampling and filtering using Python:
1. **Sampling** reduced the volume of data processed from 100 elements to 20 elements (20% retention). This demonstrates how sampling reduces memory and computation bottlenecks without introducing value-based bias.
2. **Filtering** extracted only relevant data points meeting a specific criterion (values > 50), demonstrating how condition-based checks allow real-time systems to extract meaningful signals from noisy streams.
3. Together, sampling and filtering form fundamental building blocks for scalable stream data processing pipelines.

### Exercise 4: Conceptual Difference Between Sampling and Filtering

The key conceptual differences between stream sampling and stream filtering are:

1. **Criterion of Selection**:
   - **Sampling** is non-selective regarding the content of the data; it treats every incoming element with equal probability (uniform random sampling).
   - **Filtering** is strictly conditional; it inspects the data payload and tests whether it satisfies a specified rule (e.g., threshold crossing).

2. **Control Over Volume**:
   - **Sampling** gives direct control over output size or rate (e.g., exactly $k$ elements or a fixed sampling fraction $p$).
   - **Filtering** produces an output size dependent entirely on the underlying data distribution; if all items meet the criteria, no volume reduction occurs.

3. **Information Loss vs. Noise Removal**:
   - **Sampling** causes statistical approximation; rare anomalies may be accidentally dropped.
   - **Filtering** removes known irrelevant data points (noise) while guaranteeing that all events of interest are captured.

4. **Summary Comparison Table**:
| Dimension | Sampling | Filtering |
| :--- | :--- | :--- |
| **Input Evaluation** | Blind / content-agnostic | Conditional / content-aware |
| **Output Size** | Fixed & predictable | Variable & data-dependent |
| **Primary Use** | Load shedding, approximation | Event detection, noise reduction |


### Exercise 5: Apply Sampling and Filtering to Another Numerical Dataset
We apply stream sampling and filtering to an environmental monitoring dataset: simulated hourly temperature sensor readings (°C) over 50 hours (values between 15.0°C and 45.0°C).
- **Sampling**: Select 10 random readings for routine sensor health telemetry.
- **Filtering**: Extract temperature readings exceeding 38.0°C to trigger critical heat warning alerts.


```python
# Exercise 5: Simulated Hourly Temperature Sensor Readings (°C)
sensor_temperatures = [round(random.uniform(15.0, 45.0), 1) for _ in range(50)]

print("Original Temperature Stream (50 readings in °C):")
print(sensor_temperatures)

# 1. Random Sampling: Select 10 readings for routine telemetry
sampled_temperatures = random.sample(sensor_temperatures, 10)
print("\nSampled Telemetry Readings (10 samples):")
print(sampled_temperatures)

# 2. Filtering: Extract critical heat alerts (> 38.0 °C)
heat_alerts = [temp for temp in sensor_temperatures if temp > 38.0]
print("\nFiltered Critical Heat Alerts (Readings > 38.0 °C):")
print(heat_alerts)

# Summary comparison DataFrame using Pandas and NumPy
sensor_summary_df = pd.DataFrame({
    "Stream Category": [
        "Full Sensor Stream",
        "Sampled Telemetry (Routine Check)",
        "Filtered Alerts (Critical Temp > 38°C)"
    ],
    "Reading Count": [
        len(sensor_temperatures),
        len(sampled_temperatures),
        len(heat_alerts)
    ],
    "Retention Rate (%)": [
        "100.00%",
        f"{(len(sampled_temperatures) / len(sensor_temperatures)) * 100:.2f}%",
        f"{(len(heat_alerts) / len(sensor_temperatures)) * 100:.2f}%"
    ],
    "Mean Temperature (°C)": [
        round(float(np.mean(sensor_temperatures)), 2),
        round(float(np.mean(sampled_temperatures)), 2),
        round(float(np.mean(heat_alerts)) if heat_alerts else 0.0, 2)
    ]
})

print("\n=== Temperature Sensor Stream Analysis ===")
sensor_summary_df
```
