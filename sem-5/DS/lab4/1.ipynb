# Statistical Analysis on Titanic Dataset
Loading libraries, dataset, and handling missing values.

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv('titanic.csv')
df.columns = df.columns.str.lower()

df['fare'] = df['fare'].fillna(df['fare'].median())
df['age'] = df['age'].fillna(df['age'].median())
if 'embarked' in df.columns:
    df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

print("Dataset loaded. Shape:", df.shape)
```

## 1. Perform a t-test using the Fare variable instead of Age
We will perform an Independent Two-Sample t-test to check if there is a significant difference in the average `fare` paid by passengers who survived versus those who did not.

```python
fare_survived = df[df['survived'] == 1]['fare']
fare_died = df[df['survived'] == 0]['fare']
# this is welch's t-test where we dont assume that both the groups have equal variance
# 
t_stat, p_val_ttest = stats.ttest_ind(fare_survived, fare_died, equal_var=False)
print(f"T test value: {t_stat:.4f}")
print(f"P-value: {p_val_ttest:.4e}")
```

## 2 & 3. Change significance level to 0.01 and Interpret the p-value
- **Significance level (alpha):** 0.01
- **Null Hypothesis (H0):** There is no significant difference in the mean fare between passengers who survived and those who died.
- **Alternative Hypothesis (H1):** There is a significant difference in the mean fare between the two groups.

```python
alpha = 0.01
print(f"Significance level (alpha) = {alpha}")
print(f"P-value = {p_val_ttest:.4e}")

if p_val_ttest < alpha:
    print("\nConclusion: Since p-value < 0.01, we REJECT the null hypothesis.")
    print("Interpretation: There is a statistically significant difference in the fare paid by survivors vs non-survivors.")
else:
    print("\nConclusion: Since p-value >= 0.01, we FAIL TO REJECT the null hypothesis.")
    print("Interpretation: There is NO statistically significant difference in the fare paid by survivors vs non-survivors.")
```

## 4 & 5. Apply another statistical test (Chi-Square) and state hypotheses for a different variable
We will apply the Chi-Square Test of Independence on two categorical variables: `pclass` (Passenger Class) and `survived`.

**Hypotheses:**
- **Null Hypothesis (H0):** Passenger Class (`pclass`) and Survival (`survived`) are strictly independent (no relationship).
- **Alternative Hypothesis (H1):** There is a significant relationship/dependency between Passenger Class and Survival.

```python
contingency_table = pd.crosstab(df['pclass'], df['survived'])
print("Contingency Table (Pclass vs Survived):\n")
print(contingency_table)
# dof = number of rows - 1 * number of columns - 1
# expected is the count we would expect if the two groups have no relationship
chi2_stat, p_val_chi2, dof, expected = stats.chi2_contingency(contingency_table)

print(f"\nChi-Square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_val_chi2:.4e}")
print(f"Degrees of Freedom: {dof}")

if p_val_chi2 < 0.01:
    print("\nConclusion: Since p-value < 0.01, we REJECT the null hypothesis.")
    print("Interpretation: Passenger Class has a statistically significant effect on Survival rate.")
else:
    print("\nConclusion: Since p-value >= 0.01, we FAIL TO REJECT the null hypothesis.")
```
